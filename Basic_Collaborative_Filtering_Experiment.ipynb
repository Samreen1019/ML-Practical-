# Experiment: Collaborative Filtering Recommendation System

**Aim:** To implement user-based and item-based collaborative filtering using cosine similarity and recommend top-N items to a user.

**Author: Dr. Priya B. Pandharbale**

**Libraries:** Pandas, NumPy and Scikit-learn

Run the cells one by one.

## 1. Import libraries and create rating data

```python
import pandas as pd
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

# Rows are users and columns are movies
ratings = pd.DataFrame({
    "Movie A": [5, 3, 0, 0, 2],
    "Movie B": [4, 5, 0, 3, 0],
    "Movie C": [0, 0, 5, 4, 3],
    "Movie D": [0, 2, 4, 5, 5],
    "Movie E": [3, 0, 0, 0, 4]
}, index=["User 1", "User 2", "User 3", "User 4", "User 5"])

print("Movie Rating Table:")
print(ratings)
```

**Note:** A value of `0` means that the user has not rated that movie.

## 2. User-based Collaborative Filtering

```python
# Find similarity between users
user_similarity = cosine_similarity(ratings)

user_similarity = pd.DataFrame(
    user_similarity,
    index=ratings.index,
    columns=ratings.index
)

print("User Similarity:")
print(user_similarity.round(3))
```

### Recommend movies to User 1 using similar users

```python
# Select target user
user = "User 1"

# Find similar users
similar_users = user_similarity[user].drop(user)

# Calculate recommendation scores
other_users_ratings = ratings.drop(index=user)
scores = other_users_ratings.T.dot(similar_users)

# Select movies not rated by User 1
unrated = ratings.loc[user] == 0
scores = scores[unrated]

# Display top 3 recommendations
print("Top 3 User-based Recommendations:")
print(scores.sort_values(ascending=False).head(3))
```

## 3. Item-based Collaborative Filtering

```python
# Find similarity between movies
item_similarity = cosine_similarity(ratings.T)

item_similarity = pd.DataFrame(
    item_similarity,
    index=ratings.columns,
    columns=ratings.columns
)

print("Item-Based Movie Similarity Matrix:")
print(item_similarity.round(3))
```

### Recommend movies to User 1 using similar movies

```python
# Ratings given by User 1
user_ratings = ratings.loc["User 1"]

# Calculate recommendation scores for each movie
scores = item_similarity.dot(user_ratings)

# Select movies not rated by User 1
unrated = user_ratings == 0
scores = scores[unrated]

# Display top 3 recommendations
print("Top 3 Item-Based Recommendations:")
print(scores.sort_values(ascending=False).head(3))
```

Observations
1. Collaborative filtering recommends movies based on user preferences.
2. User-based filtering identifies users having similar rating patterns.
3. Item-based filtering identifies movies having similar rating patterns.
4. Cosine similarity is used to measure similarity between users and movies.
5. Top-N recommendations are generated from movies that the user has not rated.
6. Both user-based and item-based methods recommended Movie D and Movie C for User 1.
7. Movie D received a higher recommendation score in both methods.


Result
User-based and item-based collaborative filtering were successfully implemented using cosine similarity. The similarity between users and movies was calculated, and top movie recommendations were generated for User 1. Both methods recommended Movie D and Movie C, with Movie D receiving the highest recommendation score.
